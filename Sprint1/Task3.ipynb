<h2 style="color:DodgerBlue">Название проекта:</h2>

----

### Вариант задания № 3

**Сотрудники компании (Employee)** - иерархия классов сотрудников с конструкторами, свойствами (геттерами и сеттерами) и взаимодействием объектов между собой.

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Необходимо доработать базовый класс Employee и производные классы Manager, Developer, Designer: хранить данные в закрытых полях, открыть доступ к ним через свойства с геттерами и сеттерами, добавить конструкторы классов и реализовать взаимодействие объектов между собой.

<h3 style="color:DodgerBlue">Базовый класс Employee</h3>

**Поля и свойства:**
`Name` - имя сотрудника;
`Age` - возраст;
`Salary` - зарплата, сеттер не принимает отрицательное значение.

**Конструктор:**
`Employee(string name, int age, int salary)` - задает начальные значения через свойства.

**Методы:**
`GetInfo()` - возвращает информацию о сотруднике в виде строки;
`Work()` - выводит сообщение о выполнении рабочих обязанностей;
`TakeVacation()` - выводит сообщение о взятии отпуска;
`TakeTask()` - сотрудник принимает задачу от менеджера.

**Производные классы**

Класс | Дополнительное свойство | Конструктор | Переопределяемый метод

`Manager` (Менеджер) | `SubordinatesCount` - количество подчиненных | `Manager(name, age, salary, subordinatesCount)` | `Work()` - добавляется информация об управлении командой

`Developer` (Разработчик) | `Specialization` - специализация | `Developer(name, age, salary, specialization)` | `TakeVacation()` - добавляется информация о согласовании отпуска с проектными сроками

`Designer` (Дизайнер) | `DesignStyle` - стиль дизайна | `Designer(name, age, salary, designStyle)` | `Work()` - добавляется информация о работе над дизайн-проектами

**Взаимодействие объектов:**
метод `GiveTask(Employee employee, string task)` класса `Manager` обращается к другому объекту и вызывает у него метод `TakeTask()`.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
Employee employee = new Employee("Иван Иванов", 35, 60000);
Manager manager = new Manager("Андрей Мирный", 42, 120000, 7);
Developer developer = new Developer("Арсений Календарёв", 25, 150000, "Backend");
Designer designer = new Designer("Дмитрий Кучинский", 29, 95000, "минимализм");

Console.WriteLine(employee.GetInfo());
Console.WriteLine(manager.GetInfo());
Console.WriteLine(developer.GetInfo());
Console.WriteLine(designer.GetInfo());
Console.WriteLine();

manager.GiveTask(developer, "написать модуль отчетов");
manager.GiveTask(designer, "нарисовать макет главной страницы");
Console.WriteLine();

manager.Work();
developer.Work();
developer.WriteCode();
designer.Work();
designer.DrawMockup();
Console.WriteLine();

developer.TakeVacation();
designer.Salary = -5000;
manager.Salary = 130000;
Console.WriteLine(manager.GetInfo());

class Employee
{
    private string name = "";
    private int age;
    private int salary;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public int Age
    {
        get { return age; }
        set { age = value; }
    }

    public int Salary
    {
        get { return salary; }
        set
        {
            if (value >= 0)
            {
                salary = value;
            }
            else
            {
                Console.WriteLine("Зарплата не может быть отрицательной");
            }
        }
    }

    public Employee(string name, int age, int salary)
    {
        Name = name;
        Age = age;
        Salary = salary;
    }

    public string GetInfo()
    {
        return $"{Name}, {Age} лет, зарплата {Salary} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет рабочие обязанности.");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void TakeTask(string task)
    {
        Console.WriteLine($"{Name} принял задачу: {task}");
    }
}

class Manager : Employee
{
    private int subordinatesCount;

    public int SubordinatesCount
    {
        get { return subordinatesCount; }
        set { subordinatesCount = value; }
    }

    public Manager(string name, int age, int salary, int subordinatesCount) : base(name, age, salary)
    {
        SubordinatesCount = subordinatesCount;
    }

    public void GiveTask(Employee employee, string task)
    {
        Console.WriteLine($"{Name} выдает задачу сотруднику {employee.Name}");
        employee.TakeTask(task);
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Управляет командой из {SubordinatesCount} человек.");
    }
}

class Developer : Employee
{
    private string specialization = "";

    public string Specialization
    {
        get { return specialization; }
        set { specialization = value; }
    }

    public Developer(string name, int age, int salary, string specialization) : base(name, age, salary)
    {
        Specialization = specialization;
    }

    public void WriteCode()
    {
        Console.WriteLine($"{Name} пишет код ({Specialization}).");
    }

    public override void TakeVacation()
    {
        base.TakeVacation();
        Console.WriteLine("Отпуск необходимо согласовать с проектными сроками.");
    }
}

class Designer : Employee
{
    private string designStyle = "";

    public string DesignStyle
    {
        get { return designStyle; }
        set { designStyle = value; }
    }

    public Designer(string name, int age, int salary, string designStyle) : base(name, age, salary)
    {
        DesignStyle = designStyle;
    }

    public void DrawMockup()
    {
        Console.WriteLine($"{Name} рисует макет.");
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Работает над дизайн-проектами в стиле {DesignStyle}.");
    }
}
```
