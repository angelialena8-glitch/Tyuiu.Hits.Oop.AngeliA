<h1 style="color:DodgerBlue">Индивидальный проект</h1>

<h2 style="color:DodgerBlue">Название проекта:</h2>

----

### Вариант задания № 3

**Сотрудники компании (Employee)** - иерархия классов сотрудников с новыми атрибутами и методами, демонстрирующая простое, сложное и множественное наследование.

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Необходимо доработать базовый класс Employee и производные классы Manager, Developer, Designer: добавить к ним новые атрибуты и методы, а также реализовать простое наследование (Employee → Manager, Developer, Designer), сложное наследование (Employee → Developer → TeamLead) и множественное наследование через интерфейсы IReportable и IMentor.

<h3 style="color:DodgerBlue">Базовый класс Employee</h3>

**Поля и свойства:**
`Name` - имя сотрудника;
`Age` - возраст;
`Salary` - зарплата, сеттер не принимает отрицательное значение;
`Position` - должность (новое);
`Experience` - стаж работы, сеттер не принимает отрицательное значение (новое);
`Department` - отдел (новое);
`Count` - статический счетчик созданных сотрудников (новое).

**Конструктор:**
`Employee(string name, int age, int salary, int experience, string department)` - задает начальные значения через свойства.

**Методы:**
`GetInfo()` - возвращает информацию о сотруднике в виде строки;
`Work()` - выводит сообщение о выполнении рабочих обязанностей;
`TakeVacation()` - выводит сообщение о взятии отпуска;
`TakeTask()` - сотрудник получает задачу от менеджера;
`RaiseSalary(int percent)` - повышает зарплату на заданный процент (новый);
`CalculateAnnualSalary()` - возвращает годовой доход (новый);
`CelebrateBirthday()` - увеличивает возраст сотрудника на 1 (новый).

<h3 style="color:DodgerBlue">Простое наследование</h3>

| Класс | Новые свойства | Новые методы | Переопределяемые методы |
|---|---|---|---|
| `Manager` (Менеджер) | `Budget` - бюджет, сеттер не принимает отрицательное значение; `MeetingsCount` - количество проведенных совещаний; `ProjectName` - название проекта | `HoldMeeting()`, `HireEmployee(Employee)`, `ApproveVacation(Employee)`, `MakeReport()` | `GetInfo()`, `Work()` |
| `Developer` (Разработчик) | `Language` - язык программирования; `CompletedTasks` - количество выполненных задач; `Level` - уровень (Junior, Middle, Senior) | `FixBug(string)`, `CompleteTask()`, `LevelUp()` | `GetInfo()`, `TakeVacation()` |
| `Designer` (Дизайнер) | `Tool` - графический редактор; `CompletedProjects` - количество завершенных проектов; `Rating` - рейтинг от 1 до 5, сеттер проверяет диапазон | `PresentDesign(string)`, `CompleteProject()`, `ChangeStyle(string)` | `GetInfo()`, `Work()` |

Свойства и методы из прошлого спринта сохранены: `SubordinatesCount` и `GiveTask()` у `Manager`, `Specialization` и `WriteCode()` у `Developer`, `DesignStyle` и `DrawMockup()` у `Designer`.

**Конструкторы:**
`Manager(name, age, salary, experience, subordinatesCount, budget, projectName)`;
`Developer(name, age, salary, experience, specialization, language, level)`;
`Designer(name, age, salary, experience, designStyle, tool, rating)` - вызывают конструктор базового класса через `base(...)`.

<h3 style="color:DodgerBlue">Сложное наследование</h3>

`TeamLead` (Тимлид) наследуется от `Developer`, который наследуется от `Employee`. Тимлид получает свойства и методы обоих классов (`Specialization`, `Language`, `Level`, `WriteCode()`, `LevelUp()`, `TakeVacation()` и т.д.), добавляет свойство `TeamSize` - размер команды и метод `ReviewCode(Developer)`, а также переопределяет `GetInfo()` и `Work()`.

<h3 style="color:DodgerBlue">Множественное наследование</h3>

В C# класс может наследоваться только от одного класса, поэтому множественное наследование реализовано через интерфейсы:
`IReportable` - метод `MakeReport()` для формирования отчета;
`IMentor` - свойство `MenteesCount` и метод `Mentor(Employee)` для наставничества.

`Manager : Employee, IReportable` - менеджер является сотрудником и формирует отчеты;
`TeamLead : Developer, IMentor, IReportable` - тимлид является разработчиком, наставником и формирует отчеты.

**Взаимодействие объектов:**
методы `GiveTask()`, `HireEmployee()`, `ApproveVacation()` класса `Manager`, а также `ReviewCode()` и `Mentor()` класса `TeamLead` обращаются к другим объектам и вызывают их методы.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
Employee employee = new Employee("Иван Иванов", 35, 60000, 8, "Бухгалтерия");
Manager manager = new Manager("Андрей Мирный", 42, 120000, 15, 7, 500000, "Корпоративный портал");
Developer developer = new Developer("Арсений Календарёв", 25, 150000, 3, "Backend", "C#", "Junior");
Designer designer = new Designer("Дмитрий Кучинский", 29, 95000, 5, "минимализм", "Figma", 4);
TeamLead teamLead = new TeamLead("Сергей Петров", 34, 200000, 10, "Backend", "C#", 5);

Console.WriteLine("Простое наследование (Employee -> Manager, Developer, Designer):");
Console.WriteLine(employee.GetInfo());
Console.WriteLine(manager.GetInfo());
Console.WriteLine(developer.GetInfo());
Console.WriteLine(designer.GetInfo());
Console.WriteLine();

manager.Work();
developer.Work();
developer.WriteCode();
designer.Work();
designer.DrawMockup();
Console.WriteLine();

Console.WriteLine("Новые методы базового класса Employee:");
developer.RaiseSalary(10);
Console.WriteLine($"{designer.Name}: годовой доход {designer.CalculateAnnualSalary()} руб.");
employee.CelebrateBirthday();
employee.Experience = -1;
Console.WriteLine();

Console.WriteLine("Новые методы производных классов:");
manager.HoldMeeting();
manager.GiveTask(developer, "написать модуль отчетов");
developer.CompleteTask();
developer.LevelUp();
manager.GiveTask(designer, "нарисовать макет главной страницы");
designer.PresentDesign("Главная страница");
designer.CompleteProject();
designer.ChangeStyle("флэт");
designer.Rating = 7;
manager.HireEmployee(employee);
manager.ApproveVacation(designer);
Console.WriteLine();

Console.WriteLine("Сложное наследование (Employee -> Developer -> TeamLead):");
Console.WriteLine(teamLead.GetInfo());
teamLead.Work();
teamLead.WriteCode();
teamLead.ReviewCode(developer);
teamLead.CompleteTask();
teamLead.LevelUp();
teamLead.TakeVacation();
Console.WriteLine();

Console.WriteLine("Множественное наследование (класс + интерфейсы IMentor, IReportable):");
IMentor mentor = teamLead;
mentor.Mentor(developer);
mentor.Mentor(designer);
Console.WriteLine($"Учеников у наставника: {mentor.MenteesCount}");
manager.HoldMeeting();

IReportable[] reports = { manager, teamLead };
foreach (IReportable report in reports)
{
    Console.WriteLine(report.MakeReport());
}
Console.WriteLine();

Console.WriteLine($"Всего сотрудников: {Employee.Count}");

interface IReportable
{
    string MakeReport();
}

interface IMentor
{
    int MenteesCount { get; }
    void Mentor(Employee employee);
}

class Employee
{
    private string name = "";
    private int age;
    private int salary;
    private string position = "";
    private int experience;
    private string department = "";

    public static int Count { get; private set; }

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public int Age
    {
        get { return age; }
        set { age = value; }
    }

    public int Salary
    {
        get { return salary; }
        set
        {
            if (value >= 0)
            {
                salary = value;
            }
            else
            {
                Console.WriteLine("Зарплата не может быть отрицательной");
            }
        }
    }

    public string Position
    {
        get { return position; }
        set { position = value; }
    }

    public int Experience
    {
        get { return experience; }
        set
        {
            if (value >= 0)
            {
                experience = value;
            }
            else
            {
                Console.WriteLine("Стаж не может быть отрицательным");
            }
        }
    }

    public string Department
    {
        get { return department; }
        set { department = value; }
    }

    public Employee(string name, int age, int salary, int experience, string department)
    {
        Name = name;
        Age = age;
        Salary = salary;
        Experience = experience;
        Department = department;
        Position = "Сотрудник";
        Count++;
    }

    public virtual string GetInfo()
    {
        return $"{Position} {Name}, возраст {Age}, стаж {Experience} г., отдел: {Department}, зарплата {Salary} руб.";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет рабочие обязанности.");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} уходит в отпуск.");
    }

    public void TakeTask(string task)
    {
        Console.WriteLine($"{Name} получает задачу: {task}");
    }

    public void RaiseSalary(int percent)
    {
        Salary = Salary + Salary * percent / 100;
        Console.WriteLine($"{Name}: зарплата повышена на {percent}%, теперь {Salary} руб.");
    }

    public int CalculateAnnualSalary()
    {
        return Salary * 12;
    }

    public void CelebrateBirthday()
    {
        Age++;
        Console.WriteLine($"{Name} празднует день рождения! Возраст: {Age}");
    }
}

class Manager : Employee, IReportable
{
    private int subordinatesCount;
    private int budget;
    private int meetingsCount;
    private string projectName = "";

    public int SubordinatesCount
    {
        get { return subordinatesCount; }
        set { subordinatesCount = value; }
    }

    public string ProjectName
    {
        get { return projectName; }
        set { projectName = value; }
    }

    public int Budget
    {
        get { return budget; }
        set
        {
            if (value >= 0)
            {
                budget = value;
            }
            else
            {
                Console.WriteLine("Бюджет не может быть отрицательным");
            }
        }
    }

    public int MeetingsCount
    {
        get { return meetingsCount; }
    }

    public Manager(string name, int age, int salary, int experience, int subordinatesCount, int budget, string projectName)
        : base(name, age, salary, experience, "Управление")
    {
        Position = "Менеджер";
        SubordinatesCount = subordinatesCount;
        Budget = budget;
        ProjectName = projectName;
    }

    public void GiveTask(Employee employee, string task)
    {
        Console.WriteLine($"{Name} выдает задачу сотруднику {employee.Name}");
        employee.TakeTask(task);
    }

    public void HoldMeeting()
    {
        meetingsCount++;
        Console.WriteLine($"{Name} проводит совещание №{meetingsCount}");
    }

    public void HireEmployee(Employee employee)
    {
        SubordinatesCount++;
        Console.WriteLine($"{Name} берет в команду сотрудника {employee.Name}, подчиненных: {SubordinatesCount}");
    }

    public void ApproveVacation(Employee employee)
    {
        Console.WriteLine($"{Name} согласовывает отпуск сотрудника {employee.Name}");
        employee.TakeVacation();
    }

    public string MakeReport()
    {
        return $"Отчет менеджера {Name} по проекту \"{ProjectName}\": подчиненных {SubordinatesCount}, совещаний {MeetingsCount}, бюджет {Budget} руб.";
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", подчиненных {SubordinatesCount}";
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Управляет командой из {SubordinatesCount} человек.");
    }
}

class Developer : Employee
{
    private string specialization = "";
    private string language = "";
    private int completedTasks;
    private string level = "";

    public string Specialization
    {
        get { return specialization; }
        set { specialization = value; }
    }

    public string Language
    {
        get { return language; }
        set { language = value; }
    }

    public int CompletedTasks
    {
        get { return completedTasks; }
    }

    public string Level
    {
        get { return level; }
        set { level = value; }
    }

    public Developer(string name, int age, int salary, int experience, string specialization, string language, string level)
        : base(name, age, salary, experience, "Разработка")
    {
        Position = "Разработчик";
        Specialization = specialization;
        Language = language;
        Level = level;
    }

    public void WriteCode()
    {
        Console.WriteLine($"{Name} пишет код на {Language} ({Specialization}).");
    }

    public void FixBug(string bug)
    {
        Console.WriteLine($"{Name} исправляет ошибку: {bug}");
    }

    public void CompleteTask()
    {
        completedTasks++;
        Console.WriteLine($"{Name}: задача выполнена, всего выполнено задач: {completedTasks}");
    }

    public void LevelUp()
    {
        if (Level == "Junior")
        {
            Level = "Middle";
        }
        else if (Level == "Middle")
        {
            Level = "Senior";
        }
        else
        {
            Console.WriteLine($"{Name} уже имеет максимальный уровень {Level}");
            return;
        }
        Console.WriteLine($"{Name}: уровень повышен до {Level}");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", {Level} {Specialization} ({Language})";
    }

    public override void TakeVacation()
    {
        base.TakeVacation();
        Console.WriteLine("Отпуск необходимо согласовать с проектными сроками.");
    }
}

class Designer : Employee
{
    private string designStyle = "";
    private string tool = "";
    private int completedProjects;
    private int rating;

    public string DesignStyle
    {
        get { return designStyle; }
        set { designStyle = value; }
    }

    public string Tool
    {
        get { return tool; }
        set { tool = value; }
    }

    public int CompletedProjects
    {
        get { return completedProjects; }
    }

    public int Rating
    {
        get { return rating; }
        set
        {
            if (value >= 1 && value <= 5)
            {
                rating = value;
            }
            else
            {
                Console.WriteLine("Рейтинг должен быть от 1 до 5");
            }
        }
    }

    public Designer(string name, int age, int salary, int experience, string designStyle, string tool, int rating)
        : base(name, age, salary, experience, "Дизайн")
    {
        Position = "Дизайнер";
        DesignStyle = designStyle;
        Tool = tool;
        Rating = rating;
    }

    public void DrawMockup()
    {
        Console.WriteLine($"{Name} рисует макет в {Tool}.");
    }

    public void PresentDesign(string project)
    {
        Console.WriteLine($"{Name} презентует дизайн проекта \"{project}\"");
    }

    public void CompleteProject()
    {
        completedProjects++;
        Console.WriteLine($"{Name}: дизайн-проект завершен, всего проектов: {completedProjects}");
    }

    public void ChangeStyle(string newStyle)
    {
        Console.WriteLine($"{Name} меняет стиль дизайна: {DesignStyle} -> {newStyle}");
        DesignStyle = newStyle;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", стиль: {DesignStyle}, рейтинг {Rating}";
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Работает над дизайн-проектами в стиле {DesignStyle}.");
    }
}

class TeamLead : Developer, IMentor, IReportable
{
    private int teamSize;
    private int menteesCount;

    public int TeamSize
    {
        get { return teamSize; }
        set { teamSize = value; }
    }

    public int MenteesCount
    {
        get { return menteesCount; }
    }

    public TeamLead(string name, int age, int salary, int experience, string specialization, string language, int teamSize)
        : base(name, age, salary, experience, specialization, language, "Senior")
    {
        Position = "Тимлид";
        TeamSize = teamSize;
    }

    public void ReviewCode(Developer developer)
    {
        Console.WriteLine($"{Name} проверяет код разработчика {developer.Name}");
        developer.FixBug("замечания после код-ревью");
    }

    public void Mentor(Employee employee)
    {
        menteesCount++;
        Console.WriteLine($"{Name} делится опытом с сотрудником {employee.Name}");
    }

    public string MakeReport()
    {
        return $"Отчет тимлида {Name}: команда {TeamSize} чел., выполнено задач {CompletedTasks}, учеников {MenteesCount}";
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", команда {TeamSize} чел.";
    }

    public override void Work()
    {
        base.Work();
        Console.WriteLine($"Руководит командой разработки из {TeamSize} человек.");
    }
}
```
